```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample movie dataset
data = {
    "title": ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E"],
    "budget": [100, 150, 120, 80, 200],
    "revenue": [300, 400, 250, 150, 500],
    "year": [2018, 2019, 2020, 2021, 2022],
    "genre": ["Action", "Comedy", "Action", "Drama", "Action"]
}

df = pd.DataFrame(data)

# Calculate profit
df["profit"] = df["revenue"] - df["budget"]

print("Top Revenue Movies:")
print(df.sort_values(by="revenue", ascending=False))

# Plot revenue chart
plt.figure(figsize=(6,4))
sns.barplot(x="title", y="revenue", data=df)
plt.title("Revenue by Movie")
plt.xticks(rotation=45)
plt.show()
```
